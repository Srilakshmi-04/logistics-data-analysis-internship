# Week 2 – Logistics Data Cleaning & Preprocessing
This notebook demonstrates the required logistics preprocessing workflow.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv('week2_logistics_sample_data.csv')
print(df.shape)
print(df.isnull().sum())
print('Duplicates:', df.duplicated().sum())
```

```python
df = df.drop_duplicates().copy()
num_cols = ['Distance_km','Shipment_Weight_kg','Delivery_Time_days','Transportation_Cost','Inventory_Level']
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
print(df.isnull().sum())
```

```python
Q1 = df[num_cols].quantile(0.25)
Q3 = df[num_cols].quantile(0.75)
IQR = Q3 - Q1
outlier_flags = (df[num_cols] < Q1 - 1.5*IQR) | (df[num_cols] > Q3 + 1.5*IQR)
print(outlier_flags.sum())
```

```python
scaler = MinMaxScaler()
df_scaled = df.copy()
df_scaled[num_cols] = scaler.fit_transform(df_scaled[num_cols])
print(df_scaled.head())
```

## Reflection
The pipeline handles missing values, checks duplicates, flags outliers and normalizes numerical variables before downstream logistics analysis.
