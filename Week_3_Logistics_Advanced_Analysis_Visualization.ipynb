# Week 3 – Advanced Data Analysis and Visualization in Logistics

This notebook performs exploratory data analysis (EDA), descriptive statistics, correlation analysis, visualizations, and interpretation of a hypothetical logistics dataset.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("week2_logistics_sample_data.csv")
num_cols = ["Distance_km","Shipment_Weight_kg","Delivery_Time_days",
            "Transportation_Cost","Inventory_Level"]
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
print(df.head())
print(df.shape)
```

## 1. Descriptive Statistics


```python
summary = df[num_cols].describe().round(2)
print(summary)
```

Descriptive statistics summarize central tendency and spread for key logistics variables.


## 2. Average Delivery Time by Shipment Mode


```python
mode_delivery = df.groupby("Shipment_Mode")["Delivery_Time_days"].mean().sort_values()
print(mode_delivery)
mode_delivery.plot(kind="bar", figsize=(7,4))
plt.title("Average Delivery Time by Shipment Mode")
plt.xlabel("Shipment Mode")
plt.ylabel("Average Delivery Time (days)")
plt.tight_layout()
plt.show()
```

A bar chart compares average delivery time across shipment modes.


## 3. Distribution of Delivery Times


```python
plt.figure(figsize=(7,4))
plt.hist(df["Delivery_Time_days"], bins=8)
plt.title("Distribution of Delivery Times")
plt.xlabel("Delivery Time (days)")
plt.ylabel("Number of Shipments")
plt.tight_layout()
plt.show()
```

A histogram shows the distribution and spread of delivery times and highlights longer observations.


## 4. Distance vs Transportation Cost


```python
plt.figure(figsize=(7,4))
plt.scatter(df["Distance_km"], df["Transportation_Cost"])
plt.title("Distance vs Transportation Cost")
plt.xlabel("Distance (km)")
plt.ylabel("Transportation Cost")
plt.tight_layout()
plt.show()
```

A scatter plot examines the relationship between shipment distance and transportation cost.


## 5. Correlation Analysis


```python
corr = df[num_cols].corr().round(2)
print(corr)
```

```python
plt.figure(figsize=(8,5))
plt.imshow(corr, aspect="auto")
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45, ha="right")
plt.yticks(range(len(corr.index)), corr.index)
for i in range(len(corr)):
    for j in range(len(corr.columns)):
        plt.text(j, i, f"{corr.iloc[i,j]:.2f}", ha="center", va="center")
plt.title("Correlation Matrix of Numerical Logistics Variables")
plt.colorbar(label="Correlation")
plt.tight_layout()
plt.show()
```

Correlation indicates the strength and direction of linear association. It should be interpreted as association rather than causation.


## 6. Reflection

EDA provides a structured view of delivery performance, shipment characteristics, transportation cost, and inventory levels. These findings provide a foundation for deeper predictive modeling and optimization.
